# Payoffs de posiciones largas y cortas en opciones Call


```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
K = 100
S_t = np.linspace(50, 200, 500)

# Ganancia de un call largo
ganancia_call_largo = np.maximum(S_t - K, 0)

# Ganancia de un call corto: posicion opuesta al call largo
ganancia_call_corto = -ganancia_call_largo

fig, axes = plt.subplots(1, 2, figsize=(14, 5), sharey=True)

axes[0].plot(S_t, ganancia_call_largo, color="darkgreen", linewidth=2.5)
axes[0].axhline(0, color="black", linewidth=0.8)
axes[0].axvline(K, color="gray", linestyle="--", linewidth=1.2, label=f"K = {K}")
axes[0].set_title("Ganancia: call largo")
axes[0].set_xlabel("Precio del subyacente $S_T$")
axes[0].set_ylabel("Ganancia")
axes[0].grid(alpha=0.3)
axes[0].legend()

axes[1].plot(S_t, ganancia_call_corto, color="firebrick", linewidth=2.5)
axes[1].axhline(0, color="black", linewidth=0.8)
axes[1].axvline(K, color="gray", linestyle="--", linewidth=1.2, label=f"K = {K}")
axes[1].set_title("Ganancia: call corto")
axes[1].set_xlabel("Precio del subyacente $S_T$")
axes[1].grid(alpha=0.3)
axes[1].legend()

plt.suptitle("Ganancias de opciones con precio de ejercicio K = 100")
plt.tight_layout()
plt.show()
```
