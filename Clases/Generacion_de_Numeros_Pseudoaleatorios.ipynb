Generador de ciclo no completo

Generador de ciclo completo

```python
import numpy as np
list=[]
x=7
for i in range(50):
    num = ((5*x + 3)%16)
    x=num
    list.append(num)
print(list)
```

Un generador congruencial tiene un ciclo completo si y solo si:
- m y b son primos entre ellos
- Si q es un numero primo que divide a m, entonces q divide a a-1
- Si 4 divide a m, entonces divide a a-1

```python
import numpy as np
m = 16
a = 5
b = 3
Xo1 = 0
X02 = 3
n=39

def generar_pseudoaleatorios(a, b, m, Xo, n):
    list=[]
    for i in range(n):
        num = ((a*Xo + b)%m)
        Xo = num
        list.append(num)
    return list

list1 = generar_pseudoaleatorios(a, b, m, Xo1, 10**4)
list2 = generar_pseudoaleatorios(a, b, m, X02, 10**4)

def normalizar(list):
    array = np.array(list)
    array_new = (array - min(array)) / (max(array) - min(array))
    array_new = array_new * 0.9 + 1e-9
    return array_new

u1 = normalizar(list1)
u2 = normalizar(list2)

X = np.sqrt(-2*np.log(u1)) * np.cos(2*(np.pi)*u2)
Y = np.sqrt(-2*np.log(u1)) * np.sin(2*(np.pi)*u2)

plt.hist(X)
plt.show()
```
