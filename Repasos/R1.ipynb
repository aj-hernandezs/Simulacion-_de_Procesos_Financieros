3. Paths de precios con Monte Carlo

```python
import numpy as np

np.random.seed(15)
def paths_gbm(S0, r, sigma, T, steps, n):
    dt = T/steps
    paths = np.zeros((n, steps+1))
    paths[:,0] = S0
    z = np.random.normal(0,1,(n,steps))
    for i in range(1, steps+1):
        paths[:,i] =    paths[:,i-1]*np.exp((r - 0.5*(sigma**2))*dt + sigma*np.sqrt(dt)*z[:,i-1])
    return paths

paths_10 = paths_gbm(100, 0.05, 0.2, 1, 252, 10)
paths_100 = paths_gbm(100, 0.05, 0.2, 1, 252, 100)
paths_1000 = paths_gbm(100, 0.05, 0.2, 1, 252, 1000)

'''def payoff_eu(paths, K, call=True):
    ST = paths[:,-1]
    if call:
        payoffs = np.maximum(ST - K, 0)
        ev = payoffs.mean()
    else:
        payoffs = np.maximum(K-ST, 0)
        ev = payoffs.mean()
    return ev
def payoff_as(paths, K, call=True):
    ST = []
    for i in range(paths.shape[0]):
        ST.append(paths[i,:].mean())
    ST = np.array(ST)
    if call:
        payoffs = np.maximum(ST - K, 0)
        ev = payoffs.mean()
    else:
        payoffs = np.maximum(K-ST, 0)
        ev = payoffs.mean()
    return ev
def payoff_lb(paths, call=True): #Lookback Option
    if call:
        for i in range(paths.shape[0]):
            path = paths[i,:]
            payoffs = np.maximum(path[-1] - min(path), 0)
        ev = payoffs.mean()
    else:
        for i in range(paths.shape[0]):
            path = paths[i,:]
            payoffs = np.maximum(max(path) - path[-1], 0)
        ev = payoffs.mean()
    return ev'''

def payoff_eu(paths, K, call=True):
    ST = paths[:, -1]
    payoffs = np.maximum(ST - K, 0) if call else np.maximum(K - ST, 0)
    return payoffs.mean()

def payoff_as(paths, K, call=True):
    # Usa paths[:, 1:] si no quieres incluir S0 en el promedio asiático
    average_price = paths.mean(axis=1)
    payoffs = (
        np.maximum(average_price - K, 0)
        if call
        else np.maximum(K - average_price, 0)
    )
    return payoffs.mean()

def payoff_lb(paths, call=True):
    ST = paths[:, -1]

    if call:
        payoffs = np.maximum(ST - paths.min(axis=1), 0)
    else:
        payoffs = np.maximum(paths.max(axis=1) - ST, 0)

    return payoffs.mean()

po_call_eu = payoff_eu(paths_1000, 110)
po_call_as = payoff_as(paths_1000, 110)
po_call_lb = payoff_lb(paths_1000)

print(po_call_as, po_call_eu, po_call_lb)

```
